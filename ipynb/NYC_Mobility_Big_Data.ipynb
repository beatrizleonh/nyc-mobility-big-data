```python
from pyspark.sql.functions import sum as spark_sum, when, col

calidad = df.select(
    spark_sum(when(col("trip_miles") <= 0, 1).otherwise(0)).alias("distancia_no_positiva"),
    spark_sum(when(col("trip_time") <= 0, 1).otherwise(0)).alias("duracion_no_positiva"),
    spark_sum(when(col("base_passenger_fare") < 0, 1).otherwise(0)).alias("tarifa_negativa"),
    spark_sum(when(col("driver_pay") < 0, 1).otherwise(0)).alias("pago_conductor_negativo"),
    spark_sum(when(col("dropoff_datetime") <= col("pickup_datetime"), 1).otherwise(0)).alias("fechas_invalidas")
)

calidad.show()
```

```python
df_clean = df.filter(
    (col("trip_miles") > 0) &
    (col("trip_time") > 0) &
    (col("base_passenger_fare") >= 0) &
    (col("driver_pay") >= 0) &
    (col("dropoff_datetime") > col("pickup_datetime"))
)

print("Reglas de limpieza aplicadas")
```

```python
total_limpios = df_clean.count()
eliminados = total_registros - total_limpios

print(f"Registros originales: {total_registros:,}")
print(f"Registros limpios: {total_limpios:,}")
print(f"Registros eliminados: {eliminados:,}")
print(f"Porcentaje eliminado: {eliminados/total_registros*100:.4f}%")
```

```python
from pyspark.sql.functions import year, month, dayofweek, hour, round

df_transformed = (
    df_clean
    .withColumn("year", year("pickup_datetime"))
    .withColumn("month", month("pickup_datetime"))
    .withColumn("day_of_week", dayofweek("pickup_datetime"))
    .withColumn("pickup_hour", hour("pickup_datetime"))
    .withColumn("trip_minutes", round(col("trip_time") / 60, 2))
    .withColumn(
        "avg_speed_mph",
        round(col("trip_miles") / (col("trip_time") / 3600), 2)
    )
)

print("Transformaciones creadas")
```

```python
df_transformed.select(
    "pickup_datetime",
    "trip_miles",
    "trip_time",
    "year",
    "month",
    "day_of_week",
    "pickup_hour",
    "trip_minutes",
    "avg_speed_mph"
).show(10, truncate=False)
```

```python
ruta_curated = "gs://nyc-mobility-bl-jp-2026/curated/fhvhv_trips/"

df_curated = spark.read.parquet(ruta_curated)

total_curated = df_curated.count()

print(f"Registros en CURATED: {total_curated:,}")
```

```python
df_curated.groupBy("year", "month") \
    .count() \
    .orderBy("year", "month") \
    .show(36)
```

```python
print("Spark activo:", spark.sparkContext.applicationId)
print("Executors:", spark.sparkContext._jsc.sc().getExecutorMemoryStatus().size())
```

```python
from pyspark.sql.functions import count, avg, round as spark_round

monthly_summary = (
    df_curated
    .groupBy("year", "month")
    .agg(
        count("*").alias("total_trips"),
        spark_round(avg("trip_miles"), 2).alias("avg_trip_miles"),
        spark_round(avg("trip_minutes"), 2).alias("avg_trip_minutes"),
        spark_round(avg("avg_speed_mph"), 2).alias("avg_speed_mph")
    )
    .orderBy("year", "month")
)

monthly_summary.show(36, truncate=False)
```

```python
hourly_summary = (
    df_curated
    .groupBy("pickup_hour")
    .agg(
        count("*").alias("total_trips"),
        spark_round(avg("trip_minutes"), 2).alias("avg_trip_minutes"),
        spark_round(avg("avg_speed_mph"), 2).alias("avg_speed_mph")
    )
    .orderBy("pickup_hour")
)

hourly_summary.show(24, truncate=False)
```

```python
from pyspark.sql.functions import expr

daily_summary = (
    df_curated
    .groupBy("day_of_week")
    .agg(
        count("*").alias("total_trips"),
        spark_round(avg("trip_minutes"), 2).alias("avg_trip_minutes"),
        spark_round(avg("avg_speed_mph"), 2).alias("avg_speed_mph")
    )
    .withColumn(
        "day_name",
        expr("""
            CASE day_of_week
                WHEN 1 THEN 'Sunday'
                WHEN 2 THEN 'Monday'
                WHEN 3 THEN 'Tuesday'
                WHEN 4 THEN 'Wednesday'
                WHEN 5 THEN 'Thursday'
                WHEN 6 THEN 'Friday'
                WHEN 7 THEN 'Saturday'
            END
        """)
    )
    .orderBy("day_of_week")
)

daily_summary.show(7, truncate=False)
```

```python
print("Spark activo:", spark.sparkContext.applicationId)
print("Executors:", spark.sparkContext._jsc.sc().getExecutorMemoryStatus().size())
```

```python
from pyspark.sql.functions import col

top_pickup_zones = (
    df_curated
    .groupBy("PULocationID")
    .agg(
        count("*").alias("total_trips"),
        spark_round(avg("trip_minutes"), 2).alias("avg_trip_minutes"),
        spark_round(avg("trip_miles"), 2).alias("avg_trip_miles")
    )
    .orderBy(col("total_trips").desc())
)

top_pickup_zones.show(15, truncate=False)
```

```python
top_dropoff_zones = (
    df_curated
    .groupBy("DOLocationID")
    .agg(
        count("*").alias("total_trips"),
        spark_round(avg("trip_minutes"), 2).alias("avg_trip_minutes"),
        spark_round(avg("trip_miles"), 2).alias("avg_trip_miles")
    )
    .orderBy(col("total_trips").desc())
)

top_dropoff_zones.show(15, truncate=False)
```

```python
ruta_analytics = "gs://nyc-mobility-bl-jp-2026/analytics/"

monthly_summary.write.mode("overwrite").parquet(
    ruta_analytics + "monthly_summary"
)

hourly_summary.write.mode("overwrite").parquet(
    ruta_analytics + "hourly_summary"
)

daily_summary.write.mode("overwrite").parquet(
    ruta_analytics + "daily_summary"
)

top_pickup_zones.write.mode("overwrite").parquet(
    ruta_analytics + "top_pickup_zones"
)

top_dropoff_zones.write.mode("overwrite").parquet(
    ruta_analytics + "top_dropoff_zones"
)

print("Tablas ANALYTICS guardadas correctamente")
```

```python
monthly_summary.createOrReplaceTempView("monthly_summary_sql")

print("Vista SQL creada correctamente")
```

```python
resultado_sql = spark.sql("""
    SELECT
        year,
        SUM(total_trips) AS annual_trips,
        ROUND(AVG(avg_trip_miles), 2) AS avg_trip_miles,
        ROUND(AVG(avg_trip_minutes), 2) AS avg_trip_minutes,
        ROUND(AVG(avg_speed_mph), 2) AS avg_speed_mph
    FROM monthly_summary_sql
    GROUP BY year
    ORDER BY year
""")

resultado_sql.show()
```

```python
(
    resultado_sql
    .coalesce(1)
    .write
    .mode("overwrite")
    .option("header", True)
    .csv("gs://nyc-mobility-bl-jp-2026/analytics_csv/annual_sql_summary")
)

print("Resultado SQL guardado correctamente")
```
